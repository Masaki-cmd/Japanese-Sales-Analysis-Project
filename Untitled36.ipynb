## 1. データ読み込み

まず、提供されたZIPファイルを解凍し、その中身を確認します。

```python
import zipfile
import os

zip_file_path = '/content/archive (1).zip'
extract_dir = '/content/extracted_data'

os.makedirs(extract_dir, exist_ok=True)

with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

print(f"'{zip_file_path}' を '{extract_dir}' に解凍しました。")
print("解凍されたファイル:")
for root, dirs, files in os.walk(extract_dir):
    for name in files:
        print(os.path.join(root, name))
    for name in dirs:
        print(os.path.join(root, name))
```

## 2. データクリーニングとデータ特定

### 2.1. ファイルの読み込み

まず、解凍されたCSVファイルをPandas DataFrameとして読み込みます。今回は2種類のデータセットが見つかりました。

1.  `/content/extracted_data/Updated_sales.csv`
2.  `/content/extracted_data/Sales_Data/` 内の月ごとのSalesデータ

それぞれを読み込み、内容を確認します。

```python
import pandas as pd

# Updated_sales.csv の読み込み
updated_sales_path = '/content/extracted_data/Updated_sales.csv'
try:
    df_updated_sales = pd.read_csv(updated_sales_path)
    print(f"'{updated_sales_path}' を読み込みました。")
    print("\n--- Updated_sales.csv の最初の5行 ---")
    print(df_updated_sales.head())
    print("\n--- Updated_sales.csv の情報 ---")
    df_updated_sales.info()
except Exception as e:
    print(f"'{updated_sales_path}' の読み込み中にエラーが発生しました: {e}")

# Sales_Data ディレクトリ内の月ごとのSalesデータの読み込みと結合
sales_data_dir = '/content/extracted_data/Sales_Data'
all_months_data = []

print("\n--- Sales_Data ディレクトリ内の月ごとのファイルを読み込み中 ---")
for file_name in os.listdir(sales_data_dir):
    if file_name.endswith('.csv'):
        file_path = os.path.join(sales_data_dir, file_name)
        try:
            df_month = pd.read_csv(file_path)
            all_months_data.append(df_month)
            print(f"  '{file_name}' を読み込みました。")
        except Exception as e:
            print(f"  '{file_name}' の読み込み中にエラーが発生しました: {e}")

if all_months_data:
    df_monthly_sales = pd.concat(all_months_data, ignore_index=True)
    print("\n--- すべての月ごとのSalesデータを結合しました。---")
    print("\n--- 結合された月ごとのSalesデータの最初の5行 ---")
    print(df_monthly_sales.head())
    print("\n--- 結合された月ごとのSalesデータの情報 ---")
    df_monthly_sales.info()
else:
    print("Sales_Data ディレクトリ内にCSVファイルが見つかりませんでした。")
```

### 2.2. データクリーニング

読み込んだデータフレーム`df_monthly_sales`には、いくつかクリーニングが必要な点が見受けられます。

**現状の課題:**
*   `Order ID`, `Product`, `Quantity Ordered`, `Price Each`, `Order Date`, `Purchase Address`の全てが`object`型になっている。
*   `Order ID`などに`NaN`値が存在する。
*   `df_monthly_sales.head()`を見ると、`Order ID`カラムに`'Order ID'`という文字列自体が入っている行がある可能性があり、これは複数のCSVファイルを結合した際にヘッダー行がデータとして含まれてしまったためと考えられます。

これらの問題を解決し、分析に適したデータフレームに整形します。

**クリーニング手順:**
1.  **NaN値の処理:** `Order ID`がNaNの行は不完全なデータであるため削除します。
2.  **不適切な行の削除:** 'Order Date'カラムに`'Order Date'`という文字列が含まれる行（ヘッダー行が誤ってデータとして取り込まれたもの）を削除します。
3.  **データ型の変換:**
    *   `Quantity Ordered`と`Price Each`を数値型（`int`および`float`）に変換します。
    *   `Order Date`を日付時刻型（`datetime`）に変換します。

```python
# 1. NaN値の処理
# Order IDがNaNの行を削除します。これらの行は不完全な注文データであると考えられます。
print(f"NaN値を含む前の行数: {len(df_monthly_sales)}")
df_monthly_sales.dropna(subset=['Order ID'], inplace=True)
print(f"NaN値削除後の行数: {len(df_monthly_sales)}")

# 2. 不適切な行の削除
# 'Order Date'カラムに文字列'Order Date'が含まれる行を削除します。
# これは複数のCSVを結合した際にヘッダー行がデータとして含まれてしまったものと考えられます。
original_rows = len(df_monthly_sales)
df_monthly_sales = df_monthly_sales[df_monthly_sales['Order Date'] != 'Order Date']
print(f"'Order Date'文字列を含む行削除後の行数: {len(df_monthly_sales)} (削除された行数: {original_rows - len(df_monthly_sales)})")

# 3. データ型の変換
# 'Quantity Ordered'と'Price Each'を数値型に、'Order Date'をdatetime型に変換します。
# エラーが発生した場合はNaNに変換し、後で処理します。

df_monthly_sales['Quantity Ordered'] = pd.to_numeric(df_monthly_sales['Quantity Ordered'], errors='coerce')
df_monthly_sales['Price Each'] = pd.to_numeric(df_monthly_sales['Price Each'], errors='coerce')
df_monthly_sales['Order Date'] = pd.to_datetime(df_monthly_sales['Order Date'], errors='coerce')

# 型変換後に発生したNaN（例えば、'Quantity Ordered'や'Price Each'に数値に変換できない値があった場合）を再度削除
df_monthly_sales.dropna(subset=['Quantity Ordered', 'Price Each', 'Order Date'], inplace=True)

# データ型の確認
print("\n--- クリーニング後のDataFrame情報 ---")
df_monthly_sales.info()
print("\n--- クリーニング後の最初の5行 ---")
print(df_monthly_sales.head())
```

### 2.3. 特徴量エンジニアリング（データ enrichment）

分析をさらに深めるために、既存のデータからいくつかの有用な特徴量を抽出・計算します。

**追加する特徴量:**
1.  **Sales (売上):** `Quantity Ordered` × `Price Each` で各注文の総売上を計算します。これは収益分析の基本となります。
2.  **Month (月):** `Order Date`から月情報を抽出し、月ごとの販売トレンドを把握します。
3.  **Hour (時間):** `Order Date`から時間情報を抽出し、日中のどの時間帯に売上が集中しているかを特定します。これは、マーケティング戦略や人員配置の最適化に役立ちます。
4.  **City (都市) および State (州):** `Purchase Address`から都市と州の情報を抽出し、地域ごとの販売パフォーマンスや市場特性を分析します。

```python
# 1. Sales (売上) 列の計算
df_monthly_sales['Sales'] = df_monthly_sales['Quantity Ordered'] * df_monthly_sales['Price Each']

# 2. Month (月) 列の抽出
df_monthly_sales['Month'] = df_monthly_sales['Order Date'].dt.month

# 3. Hour (時間) 列の抽出
df_monthly_sales['Hour'] = df_monthly_sales['Order Date'].dt.hour

# 4. City (都市) および State (州) 列の抽出
# 購入アドレスから都市と州を抽出する関数
def get_city_state(address):
    try:
        city_state = address.split(',')[1].strip() + ', ' + address.split(',')[2].split(' ')[1].strip()
        return city_state
    except IndexError:
        return None

df_monthly_sales['CityState'] = df_monthly_sales['Purchase Address'].apply(get_city_state)

# 必要に応じて、さらにCityとStateを分割することも可能
df_monthly_sales['City'] = df_monthly_sales['CityState'].apply(lambda x: x.split(',')[0].strip() if x else None)
df_monthly_sales['State'] = df_monthly_sales['CityState'].apply(lambda x: x.split(',')[1].strip() if x else None)

# データ型の確認と最初の5行の表示
print("\n--- 特徴量追加後のDataFrame情報 ---")
df_monthly_sales.info()
print("\n--- 特徴量追加後の最初の5行 ---")
print(df_monthly_sales.head())
```

## 3. 課題定義と仮説

ここからは、クリーニングおよび特徴量エンジニアリングが完了したデータを用いて、収益向上とコスト最適化のための具体的な課題を特定し、それに対する仮説を立てていきます。

**最終評価指標（KGI）:** 収益の最大化、コストの最適化

### 3.1. 課題の特定

データセット全体を見て、売上や利益に影響を与えている可能性のある側面を洗い出します。主な視点は以下の通りです。

*   **売上のトレンド:** 月ごと、時間ごとの売上傾向はどうなっているか？特定の時期や時間帯に売上が低い、あるいは高い理由は何なのか？
*   **製品のパフォーマンス:** どの製品が最も売れているのか？利益率の高い製品は何か？売上は多いが利益率が低い製品はないか？
*   **地域ごとのパフォーマンス:** どの都市が最も売上に貢献しているのか？地域によって販売戦略を変えるべきか？
*   **顧客の行動パターン:** どのような組み合わせで商品が購入されているか？クロスセルやアップセルの機会はないか？

これらの視点から、まずは全体的な売上の傾向を把握し、そこから具体的な課題を深掘りしていきます。

### 3.4. 製品パフォーマンスの分析

次に、製品ごとの販売数と売上を分析し、収益に最も貢献している製品、あるいは改善の余地がある製品を特定します。これは、在庫管理、製品ラインナップの最適化、マーケティング戦略の方向性を決定する上で非常に重要です。

**なぜこのグラフを作成するのか:**
*   どの製品が最も売れているのか（量と金額の両面）を明確にするため。
*   製品ポートフォリオの中で、高収益の製品とそうでない製品を区別するため。
*   在庫管理の最適化や、プロモーションの対象製品を選定するための基礎情報とするため。

**仮説:**
*   特定の高価な製品が売上総額に大きく貢献している可能性がある。
*   低単価でも販売数が多い製品が、安定した売上を支えている可能性がある。
*   あまり売れていない製品がある場合、その原因を特定し、改善策を検討する必要がある。

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 製品ごとの総売上と総販売数を集計
product_sales = df_monthly_sales.groupby('Product').agg(
    total_quantity=('Quantity Ordered', 'sum'),
    total_sales=('Sales', 'sum')
).reset_index()

# 売上上位10製品
top_10_products_by_sales = product_sales.sort_values(by='total_sales', ascending=False).head(10)

# 販売数上位10製品
top_10_products_by_quantity = product_sales.sort_values(by='total_quantity', ascending=False).head(10)

print("\n--- 売上上位10製品 ---")
print(top_10_products_by_sales)
print("\n--- 販売数上位10製品 ---")
print(top_10_products_by_quantity)

# グラフの作成
fig, axes = plt.subplots(1, 2, figsize=(20, 7))

# 売上上位10製品の棒グラフ
sns.barplot(x='total_sales', y='Product', data=top_10_products_by_sales, ax=axes[0], palette='viridis')
axes[0].set_title('売上上位10製品', fontsize=16)
axes[0].set_xlabel('総売上 (USD)', fontsize=12)
axes[0].set_ylabel('製品', fontsize=12)
axes[0].ticklabel_format(style='plain', axis='x')
axes[0].grid(axis='x', linestyle='--', alpha=0.7)

# 販売数上位10製品の棒グラフ
sns.barplot(x='total_quantity', y='Product', data=top_10_products_by_quantity, ax=axes[1], palette='magma')
axes[1].set_title('販売数上位10製品', fontsize=16)
axes[1].set_xlabel('総販売数', fontsize=12)
axes[1].set_ylabel('製品', fontsize=12)
axes[1].ticklabel_format(style='plain', axis='x')
axes[1].grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### 3.5. 地域ごとの売上パフォーマンス分析

最後に、`City` (都市) ごとの売上を分析し、どの地域が売上に最も貢献しているかを特定します。これは、地域に特化したマーケティング戦略の策定、物流の最適化、または新規店舗出店の検討に役立ちます。

**なぜこのグラフを作成するのか:**
*   地域ごとの市場規模や購買力を把握するため。
*   地域特性に基づいた販売戦略（例：特定の都市でのプロモーション強化）を検討するため。
*   効率的な物流ルートの計画や、倉庫・配送センターの配置を最適化するため。

**仮説:**
*   人口の多い大都市圏が売上高で上位を占める。
*   特定の都市でなぜ売上が高いのか、その背景に何か特別な要因があるか（例：オフィス街、学生街など）。

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 都市ごとの総売上を集計
city_sales = df_monthly_sales.groupby('City')['Sales'].sum().reset_index()

# 売上順にソート
city_sales = city_sales.sort_values(by='Sales', ascending=False)

print("\n--- 都市ごとの総売上データ ---")
print(city_sales)

# グラフの作成
plt.figure(figsize=(12, 7))
sns.barplot(x='Sales', y='City', data=city_sales, palette='cubehelix')
plt.title('都市ごとの総売上', fontsize=16)
plt.xlabel('総売上 (USD)', fontsize=12)
plt.ylabel('都市', fontsize=12)
plt.ticklabel_format(style='plain', axis='x') # X軸の表記を通常の数値にする
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

## 5. 分析結果と提案

これまでのデータ分析から得られた洞察に基づき、収益向上とコスト最適化のための具体的な提案を行います。

### 5.1. 全体的な売上トレンド (月ごと)

**分析結果:**

```python
from itertools import combinations
from collections import Counter

# 同じOrder IDを持つ商品をグループ化
# NaNを削除した後に、Order IDを基準にグループ化して製品をリストにする
df_grouped = df_monthly_sales[df_monthly_sales['Order ID'].duplicated(keep=False)]

# 各注文ID内で購入された商品をリストとして取得
# 製品名を文字列として結合する際に、データ型がobjectであることを考慮し、文字列に変換して結合する
def get_product_combinations(group):
    return ','.join(group['Product'].astype(str))

products_in_order = df_grouped.groupby('Order ID').apply(get_product_combinations)

# 2つ同時に購入された商品の組み合わせを抽出
count = Counter()
for row in products_in_order:
    items = row.split(',')
    # 同じ商品が複数回リストに含まれるのを防ぐためにsetに変換してからcombinations
    # ここでは、元のデータで'Product'が重複しているケースを考慮し、uniqueな組み合わせをカウント
    for combo in combinations(sorted(list(set(items))), 2):
        count[combo] += 1

print("\n--- よく一緒に購入される商品の組み合わせ（上位10件） ---")
for key, value in count.most_common(10):
    print(f"{key}: {value}回")

# 3つ同時に購入された商品の組み合わせも見てみる
count_3 = Counter()
for row in products_in_order:
    items = row.split(',')
    for combo in combinations(sorted(list(set(items))), 3):
        count_3[combo] += 1

print("\n--- よく一緒に購入される商品の組み合わせ（3点、上位5件） ---")
for key, value in count_3.most_common(5):
    print(f"{key}: {value}回")
```
